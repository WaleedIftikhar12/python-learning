```python
import pandas as pd
import numpy as np
rng = np.random.default_rng(42)
n = 120

df_original = pd.DataFrame({
    "Student_ID": np.arange(1001, 1001 + n),
    "Department": rng.choice(
        ["Computer Science", "Business", "Data Science"], size=n
    ),
    "Study_Hours": np.round(rng.normal(3.5, 1.2, size=n).clip(0.5, 8), 1),
    "Attendance": np.round(rng.normal(82, 10, size=n).clip(40, 100), 1),
    "Assignment_Score": rng.integers(35, 101, size=n).astype(float),
    "Exam_Score": rng.integers(30, 101, size=n).astype(float),
    "Study_Method": rng.choice(
        ["Self-study", "Group study", "Online course"], size=n
    ),
    "Internet_Access": rng.choice(["Yes", "No"], size=n, p=[0.85, 0.15])
})

# Add reproducible missing values to selected columns
for col, rate in {
    "Study_Hours": 0.10,
    "Attendance": 0.08,
    "Assignment_Score": 0.12,
    "Exam_Score": 0.10,
    "Study_Method": 0.08
}.items():
    mask = rng.random(n) < rate
    df_original.loc[mask, col] = np.nan

# Add one intentionally invalid value and one exact duplicate row
df_original.loc[5, "Attendance"] = 150
df_original = pd.concat(
    [df_original, df_original.iloc[[10]]],
    ignore_index=True
)

# Save a raw copy to the notebook's current working directory
df_original.to_csv("../data/student_performance_raw.csv", index=False)

print("Raw dataset created and saved as student_performance_raw.csv")
print("Rows and columns:", df_original.shape)
display(df_original.head(10))

```
