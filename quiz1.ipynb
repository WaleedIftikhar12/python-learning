```python
sentence = input("Enter a sentence : ")

# Counters
vowels = 0
consonants = 0
spaces = 0
special_characters = 0

# Loop through every character
for char in sentence:

    if char.lower() in "aeiou":
        vowels += 1

    elif char == " ":
        spaces += 1

    elif char.isalpha():
        consonants += 1

    else:
        special_characters += 1


# Total characters
total_characters = len(sentence)

# Summary Report
print("\n----- Character Analysis -----")
print("Total Characters:", total_characters)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
print("Special Characters:", special_characters)


# Bonus Challenge
mystery = input("\nEnter your mystery letter: ")

mystery_count = 0

for char in sentence:
    if char.lower() == mystery.lower():
        mystery_count += 1

print("Mystery letter", mystery, "was found", mystery_count, "time(s).")
```
